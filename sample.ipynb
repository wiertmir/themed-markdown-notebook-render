## Mermaid: flowchart

```mermaid
flowchart LR
    A[Markdown cell] --> B{Fence language?}
    B -->|mermaid| C[Diagram]
    B -->|geojson / topojson| D[Map]
    B -->|stl| E[3D model]
    B -->|anything else| F[Code block]
```

## Mermaid: sequence

```mermaid
sequenceDiagram
    participant R as Renderer
    participant E as Extension
    R->>E: ready
    E-->>R: theme CSS
    Note over R,E: theme changes
    E-->>R: new theme CSS
```
